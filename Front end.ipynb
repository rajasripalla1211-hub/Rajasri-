```python
# ============================================================
# FRONT END — IMPORTS
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output, Markdown

print("Frontend loaded successfully.")
```

```python
# ============================================================
# FRONT END — USER INTERFACE
# ============================================================

title = widgets.HTML(
    value="""
    <div style="
        background: linear-gradient(135deg, #667eea, #764ba2);
        padding: 25px;
        border-radius: 15px;
        color: white;
        text-align: center;
        font-family: Arial;">

        <h1>📱 Screen Rotation</h1>
        <h2>Orthogonal Transformation</h2>
        <p>B.Tech Mathematics Mini Project — F6</p>
    </div>
    """
)

angle_slider = widgets.IntSlider(
    value=90,
    min=0,
    max=360,
    step=90,
    description="Angle:",
    continuous_update=False,
    style={"description_width": "70px"},
    layout=widgets.Layout(width="500px")
)

x_input = widgets.FloatText(
    value=2,
    description="X coordinate:",
    style={"description_width": "110px"}
)

y_input = widgets.FloatText(
    value=3,
    description="Y coordinate:",
    style={"description_width": "110px"}
)

calculate_button = widgets.Button(
    description="Calculate Rotation",
    button_style="primary",
    icon="refresh",
    layout=widgets.Layout(width="220px", height="40px")
)

rotate_twice_button = widgets.Button(
    description="Rotate Twice",
    button_style="info",
    icon="repeat",
    layout=widgets.Layout(width="180px", height="40px")
)

output_area = widgets.Output()

display(title)
display(
    widgets.VBox([
        widgets.HTML("<h3>⚙️ Input Settings</h3>"),
        angle_slider,
        x_input,
        y_input,
        widgets.HBox([
            calculate_button,
            rotate_twice_button
        ]),
        output_area
    ])
)
```

```python
# ============================================================
# FRONT END — CALCULATION ENGINE
# ============================================================

def frontend_calculate(angle, x, y):

    # Create rotation matrix
    Q = rotation_matrix(angle)

    # Input vector
    point = np.array([x, y], dtype=float)

    # Apply transformation
    rotated = Q @ point

    # Mathematical properties
    QTQ = Q.T @ Q
    determinant = np.linalg.det(Q)

    original_length = np.linalg.norm(point)
    rotated_length = np.linalg.norm(rotated)

    q_original = point.T @ point
    q_rotated = rotated.T @ rotated

    return (
        Q,
        point,
        rotated,
        QTQ,
        determinant,
        original_length,
        rotated_length,
        q_original,
        q_rotated
    )


def show_calculation(_=None):

    with output_area:

        clear_output(wait=True)

        angle = angle_slider.value
        x = x_input.value
        y = y_input.value

        (
            Q,
            point,
            rotated,
            QTQ,
            determinant,
            original_length,
            rotated_length,
            q_original,
            q_rotated
        ) = frontend_calculate(angle, x, y)

        display(Markdown("## 📊 Rotation Result"))

        print(f"Rotation angle = {angle}°")

        print("\nRotation Matrix Q:")
        print(Q)

        print("\nInput coordinate:")
        print(point)

        print("\nRotated coordinate:")
        print(rotated)

        print("\nQᵀQ:")
        print(QTQ)

        print("\ndet(Q):")
        print(round(determinant, 6))

        print("\nOriginal distance:")
        print(round(original_length, 6))

        print("Rotated distance:")
        print(round(rotated_length, 6))

        print("\nOriginal quadratic form xᵀx:")
        print(round(q_original, 6))

        print("Rotated quadratic form (Qx)ᵀ(Qx):")
        print(round(q_rotated, 6))

        print("\n" + "=" * 50)

        print(
            "✓ Orthogonal Matrix:",
            np.allclose(QTQ, np.eye(2))
        )

        print(
            "✓ Distance Preserved:",
            np.isclose(
                original_length,
                rotated_length
            )
        )

        print(
            "✓ Quadratic Form Preserved:",
            np.isclose(
                q_original,
                q_rotated
            )
        )
```

```python
# ============================================================
# CALCULATE BUTTON
# ============================================================

calculate_button.on_click(show_calculation)

print("Calculate button connected.")
```

```python
# ============================================================
# ROTATE TWICE
# ============================================================

def show_double_rotation(_=None):

    with output_area:

        clear_output(wait=True)

        angle = angle_slider.value

        Q = rotation_matrix(angle)

        point = np.array([
            x_input.value,
            y_input.value
        ])

        # First rotation
        first = Q @ point

        # Second rotation
        second = Q @ first

        # Equivalent single rotation
        Q_double = Q @ Q
        direct = Q_double @ point

        display(Markdown("## 🔄 Double Rotation"))

        print(f"First rotation: {angle}°")
        print(f"Second rotation: {angle}°")

        print("\nOriginal point:")
        print(point)

        print("\nAfter first rotation:")
        print(first)

        print("\nAfter second rotation:")
        print(second)

        print("\nQ²:")
        print(Q_double)

        print("\nDirect Q² transformation:")
        print(direct)

        print("\nResult:")
        print(
            "✓ Two successive rotations verified:",
            np.allclose(second, direct)
        )
```

```python
# ============================================================
# ROTATE TWICE BUTTON
# ============================================================

rotate_twice_button.on_click(show_double_rotation)

print("Rotate Twice button connected.")
```

```python
# ============================================================
# FRONT END — GRAPH
# ============================================================

def show_rotation_graph(_=None):

    with output_area:

        clear_output(wait=True)

        angle = angle_slider.value

        point = np.array([
            x_input.value,
            y_input.value
        ])

        Q = rotation_matrix(angle)

        rotated = Q @ point

        plt.figure(figsize=(8, 7))

        # Original point
        plt.scatter(
            point[0],
            point[1],
            s=150,
            label="Original Point"
        )

        # Rotated point
        plt.scatter(
            rotated[0],
            rotated[1],
            s=150,
            label=f"Rotated ({angle}°)"
        )

        # Arrow
        plt.annotate(
            "",
            xy=rotated,
            xytext=point,
            arrowprops=dict(
                arrowstyle="->",
                linewidth=2
            )
        )

        # Coordinate axes
        plt.axhline(0, linewidth=1)
        plt.axvline(0, linewidth=1)

        plt.xlabel("X")
        plt.ylabel("Y")

        plt.title(
            f"Screen Coordinate Rotation — {angle}°"
        )

        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.axis("equal")

        plt.show()
```

```python
# ============================================================
# GRAPH BUTTON
# ============================================================

graph_button = widgets.Button(
    description="Show Graph",
    button_style="success",
    icon="line-chart",
    layout=widgets.Layout(
        width="180px",
        height="40px"
    )
)

graph_button.on_click(show_rotation_graph)

display(graph_button)
```

```python
# ============================================================
# PHONE SCREEN VISUALIZATION
# ============================================================

def show_phone_rotation(_=None):

    with output_area:

        clear_output(wait=True)

        angle = angle_slider.value

        # Portrait screen
        width = 4
        height = 7

        screen = np.array([
            [-width/2, -height/2],
            [ width/2, -height/2],
            [ width/2,  height/2],
            [-width/2,  height/2],
            [-width/2, -height/2]
        ])

        Q = rotation_matrix(angle)

        rotated_screen = screen @ Q.T

        plt.figure(figsize=(9, 7))

        plt.plot(
            screen[:, 0],
            screen[:, 1],
            linewidth=3,
            label="Original Screen"
        )

        plt.plot(
            rotated_screen[:, 0],
            rotated_screen[:, 1],
            linewidth=3,
            label=f"Rotated Screen ({angle}°)"
        )

        plt.axhline(0, linewidth=1)
        plt.axvline(0, linewidth=1)

        plt.xlabel("X")
        plt.ylabel("Y")

        plt.title(
            f"📱 Phone Screen Rotation — {angle}°"
        )

        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.axis("equal")

        plt.show()
```

```python
# ============================================================
# PHONE SCREEN BUTTON
# ============================================================

phone_button = widgets.Button(
    description="Show Phone Screen",
    button_style="warning",
    icon="mobile",
    layout=widgets.Layout(
        width="220px",
        height="40px"
    )
)

phone_button.on_click(show_phone_rotation)

display(phone_button)
```

```python
# ============================================================
# FINAL PROJECT DASHBOARD
# ============================================================

dashboard_button = widgets.Button(
    description="Run Full Verification",
    button_style="danger",
    icon="check",
    layout=widgets.Layout(
        width="250px",
        height="45px"
    )
)

dashboard_output = widgets.Output()


def run_dashboard(_):

    with dashboard_output:

        clear_output(wait=True)

        angle = angle_slider.value
        Q = rotation_matrix(angle)

        orthogonal = np.allclose(
            Q.T @ Q,
            np.eye(2)
        )

        determinant = np.linalg.det(Q)

        point = np.array([
            x_input.value,
            y_input.value
        ])

        rotated = Q @ point

        distance_test = np.isclose(
            np.linalg.norm(point),
            np.linalg.norm(rotated)
        )

        q1 = point @ point
        q2 = rotated @ rotated

        quadratic_test = np.isclose(q1, q2)

        print("=" * 65)
        print("        SCREEN ROTATION PROJECT DASHBOARD")
        print("=" * 65)

        print(f"\nRotation Angle: {angle}°")

        print("\nRotation Matrix:")
        print(Q)

        print("\nMathematical Verification")
        print("-" * 65)

        print(
            f"QᵀQ = I                 : "
            f"{'PASS ✓' if orthogonal else 'FAIL ✗'}"
        )

        print(
            f"|det(Q)| = 1            : "
            f"{'PASS ✓' if np.isclose(abs(determinant), 1) else 'FAIL ✗'}"
        )

        print(
            f"Distance preserved       : "
            f"{'PASS ✓' if distance_test else 'FAIL ✗'}"
        )

        print(
            f"Quadratic form preserved : "
            f"{'PASS ✓' if quadratic_test else 'FAIL ✗'}"
        )

        print("\n" + "=" * 65)

        if all([
            orthogonal,
            np.isclose(abs(determinant), 1),
            distance_test,
            quadratic_test
        ]):
            print("🎉 ALL CHECKS PASSED")
        else:
            print("⚠️ CHECK FAILED")

        print("=" * 65)


dashboard_button.on_click(run_dashboard)

display(dashboard_button)
display(dashboard_output)
```
