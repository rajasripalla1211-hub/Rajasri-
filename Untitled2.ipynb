```python
# CELL 1: Frontend setup

!pip -q install gradio plotly

import numpy as np
import gradio as gr
import plotly.graph_objects as go
```

```python
# CELL 2: Mathematical functions

def rotation_matrix(angle):
    theta = np.radians(angle)
    return np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])

def calculate_rotation(angle, x, y):
    Q = rotation_matrix(angle)

    point = np.array([x, y])
    rotated = Q @ point

    QTQ = Q.T @ Q
    determinant = np.linalg.det(Q)

    return Q, rotated, QTQ, determinant
```

```python
# CELL 3: Create interactive screen visualization

def create_screen_plot(angle):
    theta = np.radians(angle)

    # Portrait screen
    width, height = 4, 7

    corners = np.array([
        [-width/2, -height/2],
        [ width/2, -height/2],
        [ width/2,  height/2],
        [-width/2,  height/2],
        [-width/2, -height/2]
    ])

    R = np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])

    rotated = corners @ R.T

    fig = go.Figure()

    # Original screen
    fig.add_trace(go.Scatter(
        x=corners[:, 0],
        y=corners[:, 1],
        mode="lines",
        name="Portrait Screen"
    ))

    # Rotated screen
    fig.add_trace(go.Scatter(
        x=rotated[:, 0],
        y=rotated[:, 1],
        mode="lines",
        name=f"Rotated {angle}°"
    ))

    fig.update_layout(
        title="📱 Screen Rotation",
        xaxis_title="X-axis",
        yaxis_title="Y-axis",
        xaxis=dict(scaleanchor="y"),
        template="plotly_white",
        height=500
    )

    return fig
```

```python
# CELL 4: Connect UI inputs to mathematics

def run_rotation(angle, x, y):

    Q, rotated, QTQ, determinant = calculate_rotation(angle, x, y)

    matrix_text = (
        f"Q =\n"
        f"[[{Q[0,0]:.3f}, {Q[0,1]:.3f}],\n"
        f" [{Q[1,0]:.3f}, {Q[1,1]:.3f}]]"
    )

    coordinate_text = (
        f"Original coordinate:\n({x:.2f}, {y:.2f})\n\n"
        f"Rotated coordinate:\n"
        f"({rotated[0]:.2f}, {rotated[1]:.2f})"
    )

    orthogonal_text = (
        f"QᵀQ =\n"
        f"[[{QTQ[0,0]:.3f}, {QTQ[0,1]:.3f}],\n"
        f" [{QTQ[1,0]:.3f}, {QTQ[1,1]:.3f}]]\n\n"
        f"det(Q) = {determinant:.3f}\n\n"
        f"✓ Orthogonal Matrix"
    )

    plot = create_screen_plot(angle)

    return matrix_text, coordinate_text, orthogonal_text, plot
```

```python
# CELL 5: MAIN FRONTEND ARTIFACT

custom_css = """
body {
    background: #f5f7fb;
}

.gradio-container {
    max-width: 1200px !important;
    margin: auto;
}

.title {
    text-align: center;
    font-size: 32px;
    font-weight: bold;
}

.subtitle {
    text-align: center;
    font-size: 17px;
    color: #555;
}
"""

with gr.Blocks(css=custom_css, title="Screen Rotation as Orthogonal Transformation") as demo:

    gr.Markdown(
        """
        <div class="title">
        📱 Screen Rotation as Orthogonal Transformation
        </div>

        <div class="subtitle">
        B.Tech 1st Year Mathematics Mini Project
        <br>
        Orthogonal Matrices • Rotation Matrices • Coordinate Transformation
        </div>
        """
    )

    gr.Markdown("---")

    with gr.Row():

        with gr.Column(scale=1):

            gr.Markdown("### ⚙️ Rotation Controls")

            angle = gr.Slider(
                minimum=0,
                maximum=360,
                value=90,
                step=90,
                label="Rotation Angle (°)"
            )

            x = gr.Number(
                value=2,
                label="X Coordinate"
            )

            y = gr.Number(
                value=3,
                label="Y Coordinate"
            )

            calculate_btn = gr.Button(
                "🔄 Apply Rotation",
                variant="primary"
            )

            gr.Markdown(
                """
                **Rotation options**

                • 0° → Original position
                • 90° → Portrait → Landscape
                • 180° → Upside down
                • 270° → Reverse rotation
                • 360° → Original position
                """
            )

        with gr.Column(scale=2):

            plot_output = gr.Plot(
                label="Screen Rotation Visualization"
            )

    gr.Markdown("---")

    with gr.Row():

        with gr.Column():

            gr.Markdown("### 🧮 Rotation Matrix")

            matrix_output = gr.Textbox(
                lines=4,
                interactive=False
            )

        with gr.Column():

            gr.Markdown("### 📍 Coordinate Transformation")

            coordinate_output = gr.Textbox(
                lines=5,
                interactive=False
            )

        with gr.Column():

            gr.Markdown("### ✅ Orthogonality Check")

            orthogonal_output = gr.Textbox(
                lines=7,
                interactive=False
            )

    gr.Markdown("---")

    gr.Markdown(
        """
        ### 📐 Mathematical Result

        For a rotation matrix **Q**,

        **QᵀQ = I**

        Therefore, **Q is an orthogonal matrix** and

        **det(Q) = ±1**

        Rotation preserves distances and angles.
        """
    )

    calculate_btn.click(
        fn=run_rotation,
        inputs=[angle, x, y],
        outputs=[
            matrix_output,
            coordinate_output,
            orthogonal_output,
            plot_output
        ]
    )

    # Initial display
    demo.load(
        fn=run_rotation,
        inputs=[angle, x, y],
        outputs=[
            matrix_output,
            coordinate_output,
            orthogonal_output,
            plot_output
        ]
    )

demo.launch(share=True)
```

```python
# CELL 6: Demonstrate rotating twice = 180°

def rotate_twice():

    Q90 = rotation_matrix(90)

    result = Q90 @ Q90

    return (
        "90° Rotation Matrix:\n"
        f"{np.round(Q90, 3)}\n\n"
        "Q × Q =\n"
        f"{np.round(result, 3)}\n\n"
        "Therefore:\n"
        "90° + 90° = 180°\n"
        "✓ Two rotations produce a 180° rotation."
    )

twice_output = gr.Textbox(
    label="Two 90° Rotations",
    lines=10
)

twice_button = gr.Button("🔁 Show Two Rotations")

twice_button.click(
    fn=rotate_twice,
    outputs=twice_output
)
```
