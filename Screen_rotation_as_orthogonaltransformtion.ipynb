```python
# ============================================================
# B.Tech Mathematics Mini Project
# ============================================================
# Project Title:
# Screen Rotation as Orthogonal Transformation (F6)
#
# Mathematical Concepts:
# 1. Orthogonal matrices
# 2. Rotation matrices
# 3. Quadratic form / distance invariance
#
# Environment:
# Google Colab + Python
# NumPy + SymPy + Matplotlib
# ============================================================

print("=" * 65)
print("SCREEN ROTATION AS ORTHOGONAL TRANSFORMATION")
print("=" * 65)

print("""
Objective:
Show that phone screen rotation can be represented by an
orthogonal rotation matrix.

We will:
1. Construct the portrait -> landscape rotation matrix.
2. Verify Q^T Q = I.
3. Verify det(Q) = 1.
4. Apply Q to screen coordinates.
5. Verify that distances are preserved.
6. Verify quadratic-form invariance.
7. Show that rotating twice gives a 180-degree rotation.
8. Visualize the transformation.
""")
```

```python
# ============================================================
# IMPORTS
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Make NumPy output easier to read
np.set_printoptions(precision=4, suppress=True)

print("Libraries imported successfully.")
```

```python
# ============================================================
# INPUT CONFIGURATION
# ============================================================
# Change these variables to experiment with different cases.

# Rotation angle for portrait -> landscape
ANGLE_DEGREES = 90

# Example screen size in mathematical coordinate units
SCREEN_WIDTH = 1080
SCREEN_HEIGHT = 1920

# Example point on the screen
# The point is represented as [x, y]^T
POINT = np.array([300.0, 600.0])

# A second point is useful for checking distance preservation
POINT_2 = np.array([700.0, 1000.0])

# Tolerance used for numerical verification
TOLERANCE = 1e-10

print("Input configuration")
print("-" * 40)
print(f"Rotation angle : {ANGLE_DEGREES} degrees")
print(f"Screen size    : {SCREEN_WIDTH} x {SCREEN_HEIGHT}")
print(f"Point P        : {POINT}")
print(f"Point R        : {POINT_2}")
```

```python
# ============================================================
# 90-DEGREE ROTATION MATRIX
# ============================================================

angle_rad = np.radians(ANGLE_DEGREES)

Q = np.array([
    [np.cos(angle_rad), -np.sin(angle_rad)],
    [np.sin(angle_rad),  np.cos(angle_rad)]
])

print("Rotation matrix Q:")
print(Q)

print("\nFor a 90-degree rotation, theoretically:")
print("""
        [ 0  -1 ]
Q  =    [ 1   0 ]
""")
```

```python
# ============================================================
# ORTHOGONALITY TEST
# ============================================================
# A matrix Q is orthogonal if:
#
#             Q^T Q = I
#
# where I is the identity matrix.

QT = Q.T

QTQ = QT @ Q

I = np.eye(2)

det_Q = np.linalg.det(Q)

print("Q^T:")
print(QT)

print("\nQ^T Q:")
print(QTQ)

print("\nIdentity matrix I:")
print(I)

print("\ndet(Q):")
print(det_Q)

# Numerical verification
orthogonal = np.allclose(QTQ, I, atol=TOLERANCE)
det_is_valid = np.isclose(abs(det_Q), 1.0, atol=TOLERANCE)

print("\nVerification:")
print(f"Q^T Q = I      : {orthogonal}")
print(f"|det(Q)| = 1    : {det_is_valid}")

if orthogonal and det_is_valid:
    print("\nRESULT: Q is an orthogonal matrix.")
else:
    print("\nRESULT: Orthogonality test failed.")
```

```python
# ============================================================
# APPLY ROTATION TO A SCREEN POINT
# ============================================================

P = POINT.reshape(2, 1)

P_rotated = Q @ P

print("Original point P:")
print(P)

print("\nRotated point QP:")
print(P_rotated)

print("\nCoordinate transformation:")
print(f"(x, y) = ({POINT[0]:.2f}, {POINT[1]:.2f})")

print(
    f"(x', y') = ({P_rotated[0,0]:.2f}, "
    f"{P_rotated[1,0]:.2f})"
)
```

```python
# ============================================================
# DISTANCE PRESERVATION
# ============================================================
# Orthogonal transformations preserve Euclidean distance.
#
# For vectors u and v:
#
# ||Qu - Qv|| = ||u - v||
#
# because Q^T Q = I.

P1 = POINT
P2 = POINT_2

# Original distance
original_distance = np.linalg.norm(P2 - P1)

# Rotate both points
P1_rotated = Q @ P1
P2_rotated = Q @ P2

# Distance after rotation
rotated_distance = np.linalg.norm(P2_rotated - P1_rotated)

print("Original points:")
print(f"P1 = {P1}")
print(f"P2 = {P2}")

print("\nRotated points:")
print(f"QP1 = {P1_rotated}")
print(f"QP2 = {P2_rotated}")

print("\nDistance comparison:")
print(f"Original distance = {original_distance:.10f}")
print(f"Rotated distance  = {rotated_distance:.10f}")

print(
    f"Difference        = "
    f"{abs(original_distance - rotated_distance):.10e}"
)

if np.isclose(original_distance, rotated_distance, atol=TOLERANCE):
    print("\nRESULT: Distance is preserved.")
else:
    print("\nRESULT: Distance was not preserved.")
```

```python
# ============================================================
# QUADRATIC FORM INVARIANCE
# ============================================================
#
# Original quadratic form:
#
# x^T x = x^2 + y^2
#
# After rotation:
#
# (Qx)^T(Qx)
#
# Since Q^T Q = I:
#
# (Qx)^T(Qx) = x^T Q^T Q x = x^T x

x, y = sp.symbols('x y', real=True)

vector = sp.Matrix([x, y])

# Exact 90-degree rotation matrix
Q_sym = sp.Matrix([
    [0, -1],
    [1,  0]
])

rotated_vector = Q_sym * vector

original_quadratic_form = sp.expand(
    (vector.T * vector)[0]
)

rotated_quadratic_form = sp.expand(
    (rotated_vector.T * rotated_vector)[0]
)

print("Original vector:")
sp.pprint(vector)

print("\nRotated vector:")
sp.pprint(rotated_vector)

print("\nOriginal quadratic form x^T x:")
sp.pprint(original_quadratic_form)

print("\nRotated quadratic form (Qx)^T(Qx):")
sp.pprint(rotated_quadratic_form)

print("\nDifference:")
sp.pprint(
    sp.simplify(
        rotated_quadratic_form - original_quadratic_form
    )
)
```
