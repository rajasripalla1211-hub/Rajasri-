```python
# ============================================================
# IMPORTS
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Make numerical output clean
np.set_printoptions(precision=3, suppress=True)

print("NumPy, SymPy and Matplotlib imported successfully.")
```

```python
# ============================================================
# USER CONFIGURATION
# ============================================================

# Rotation angle
ANGLE_DEGREES = 90

# Screen dimensions
SCREEN_WIDTH = 6
SCREEN_HEIGHT = 10

# Points on the screen
USER_POINTS = np.array([
    [1, 2],
    [2, 1],
    [-2, 3],
    [3, -1]
], dtype=float)

print("Project Configuration")
print("-" * 40)
print("Rotation angle :", ANGLE_DEGREES, "degrees")
print("Screen width   :", SCREEN_WIDTH)
print("Screen height  :", SCREEN_HEIGHT)

print("\nInput coordinates:")
print(USER_POINTS)
```

```python
# ============================================================
# INPUT VALIDATION
# ============================================================

if not isinstance(ANGLE_DEGREES, (int, float)):
    raise TypeError("ANGLE_DEGREES must be a number.")

if SCREEN_WIDTH <= 0 or SCREEN_HEIGHT <= 0:
    raise ValueError("Screen dimensions must be positive.")

if USER_POINTS.ndim != 2 or USER_POINTS.shape[1] != 2:
    raise ValueError("USER_POINTS must contain 2D coordinates.")

print("✓ All inputs are valid.")
```

```python
# ============================================================
# MATHEMATICAL ENGINE
# ROTATION MATRIX
# ============================================================

def rotation_matrix(angle_degrees):
    """
    Construct the standard 2D rotation matrix.

              [ cosθ  -sinθ ]
    Q(θ) =     [ sinθ   cosθ ]

    θ is measured counterclockwise.
    """

    theta = np.radians(angle_degrees)

    c = np.cos(theta)
    s = np.sin(theta)

    Q = np.array([
        [c, -s],
        [s,  c]
    ])

    # Remove extremely small floating-point errors
    Q[np.abs(Q) < 1e-12] = 0

    return Q


Q = rotation_matrix(ANGLE_DEGREES)

print("Rotation Matrix Q:")
print(Q)
```

```python
# ============================================================
# MATHEMATICAL ENGINE
# ROTATION MATRIX
# ============================================================

def rotation_matrix(angle_degrees):
    """
    Construct the standard 2D rotation matrix.

              [ cosθ  -sinθ ]
    Q(θ) =     [ sinθ   cosθ ]

    θ is measured counterclockwise.
    """

    theta = np.radians(angle_degrees)

    c = np.cos(theta)
    s = np.sin(theta)

    Q = np.array([
        [c, -s],
        [s,  c]
    ])

    # Remove extremely small floating-point errors
    Q[np.abs(Q) < 1e-12] = 0

    return Q


Q = rotation_matrix(ANGLE_DEGREES)

print("Rotation Matrix Q:")
print(Q)
```

```python
# ============================================================
# EXACT SYMBOLIC ROTATION MATRIX
# ============================================================

theta = sp.symbols('theta', real=True)

Q_symbolic = sp.Matrix([
    [sp.cos(theta), -sp.sin(theta)],
    [sp.sin(theta),  sp.cos(theta)]
])

print("General Rotation Matrix:")
sp.pprint(Q_symbolic)

# 90-degree rotation
Q_90_exact = Q_symbolic.subs(theta, sp.pi / 2)

print("\nExact 90° Rotation Matrix:")
sp.pprint(Q_90_exact)
```

```python
# ============================================================
# ORTHOGONALITY TEST
# ============================================================

Q_T = Q.T
QTQ = Q_T @ Q
I = np.eye(2)

orthogonal = np.allclose(QTQ, I)

print("Qᵀ:")
print(Q_T)

print("\nQᵀQ:")
print(QTQ)

print("\nIdentity Matrix I:")
print(I)

print("\nOrthogonal Matrix Test:")
print("QᵀQ = I :", orthogonal)
```

```python
# ============================================================
# DETERMINANT TEST
# ============================================================

det_Q = np.linalg.det(Q)

print("Rotation Matrix Q:")
print(Q)

print("\ndet(Q) =", det_Q)

print("\nOrthogonal determinant condition:")
print("|det(Q)| = 1 :", np.isclose(abs(det_Q), 1))

print("\nFor a pure rotation:")
print("det(Q) = 1")
```

```python
# ============================================================
# APPLY ROTATION TO SCREEN COORDINATES
# ============================================================

def rotate_points(points, Q):
    """
    Apply transformation:

        x' = Qx

    to a collection of 2D points.
    """

    return points @ Q.T


rotated_points = rotate_points(USER_POINTS, Q)

print("Original Coordinates:")
print(USER_POINTS)

print("\nRotated Coordinates:")
print(rotated_points)
```

```python
# ============================================================
# OUTPUT TABLE
# ============================================================

print("=" * 65)
print("COORDINATE TRANSFORMATION")
print("=" * 65)

print(f"{'Point':<10}{'Original':<20}{'Rotated':<20}")
print("-" * 65)

for i, (original, rotated) in enumerate(
    zip(USER_POINTS, rotated_points), start=1
):

    original_text = f"({original[0]:.2f}, {original[1]:.2f})"
    rotated_text = f"({rotated[0]:.2f}, {rotated[1]:.2f})"

    print(
        f"{i:<10}"
        f"{original_text:<20}"
        f"{rotated_text:<20}"
    )
```

```python
# ============================================================
# DISTANCE PRESERVATION
# ============================================================

original_lengths = np.linalg.norm(USER_POINTS, axis=1)
rotated_lengths = np.linalg.norm(rotated_points, axis=1)

lengths_preserved = np.allclose(
    original_lengths,
    rotated_lengths
)

print("=" * 65)
print("DISTANCE PRESERVATION")
print("=" * 65)

print(f"{'Point':<10}{'Before':<20}{'After':<20}")
print("-" * 65)

for i, (before, after) in enumerate(
    zip(original_lengths, rotated_lengths), start=1
):

    print(
        f"{i:<10}"
        f"{before:<20.6f}"
        f"{after:<20.6f}"
    )

print("\nDistances preserved:", lengths_preserved)
```

```python
# ============================================================
# QUADRATIC FORM INVARIANCE
# ============================================================

def quadratic_form(point):
    """
    q(x) = xᵀx

    This represents the squared Euclidean distance
    from the origin.
    """

    return point.T @ point


original_q = np.array([
    quadratic_form(p)
    for p in USER_POINTS
])

rotated_q = np.array([
    quadratic_form(p)
    for p in rotated_points
])

quadratic_invariant = np.allclose(
    original_q,
    rotated_q
)

print("=" * 65)
print("QUADRATIC FORM INVARIANCE")
print("=" * 65)

print(f"{'Point':<10}{'q(x)':<20}{'q(Qx)':<20}")
print("-" * 65)

for i, (q1, q2) in enumerate(
    zip(original_q, rotated_q), start=1
):

    print(
        f"{i:<10}"
        f"{q1:<20.6f}"
        f"{q2:<20.6f}"
    )

print("\nQuadratic form preserved:", quadratic_invariant)
```

```python
# ============================================================
# TWO SUCCESSIVE 90° ROTATIONS
# ============================================================

Q_90 = rotation_matrix(90)
Q_180 = rotation_matrix(180)

Q_twice = Q_90 @ Q_90

print("Q(90°):")
print(Q_90)

print("\nQ(90°) × Q(90°):")
print(Q_twice)

print("\nQ(180°):")
print(Q_180)

double_rotation_verified = np.allclose(
    Q_twice,
    Q_180
)

print(
    "\nQ(90°)² = Q(180°):",
    double_rotation_verified
)
```

```python
# ============================================================
# APPLY TWO ROTATIONS TO POINTS
# ============================================================

points_after_90 = rotate_points(USER_POINTS, Q_90)

points_after_180 = rotate_points(
    points_after_90,
    Q_90
)

direct_180 = rotate_points(
    USER_POINTS,
    Q_180
)

print("Original:")
print(USER_POINTS)

print("\nAfter 90°:")
print(points_after_90)

print("\nAfter two 90° rotations:")
print(points_after_180)

print("\nDirect 180° rotation:")
print(direct_180)

print(
    "\nTwo 90° rotations = one 180° rotation:",
    np.allclose(points_after_180, direct_180)
)
```

```python
# ============================================================
# PORTRAIT → LANDSCAPE VISUALIZATION
# ============================================================

w = SCREEN_WIDTH / 2
h = SCREEN_HEIGHT / 2

# Portrait screen centered at origin
portrait = np.array([
    [-w, -h],
    [ w, -h],
    [ w,  h],
    [-w,  h],
    [-w, -h]
])

# Rotate portrait screen by 90°
landscape = portrait @ Q_90.T

plt.figure(figsize=(12, 5))

# ---------------- Portrait ----------------
plt.subplot(1, 2, 1)

plt.plot(
    portrait[:, 0],
    portrait[:, 1],
    linewidth=3
)

plt.fill(
    portrait[:, 0],
    portrait[:, 1],
    alpha=0.2
)

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.title("Portrait Orientation")
plt.xlabel("x")
plt.ylabel("y")

plt.grid(True, alpha=0.3)
plt.axis("equal")

# ---------------- Landscape ----------------
plt.subplot(1, 2, 2)

plt.plot(
    landscape[:, 0],
    landscape[:, 1],
    linewidth=3
)

plt.fill(
    landscape[:, 0],
    landscape[:, 1],
    alpha=0.2
)

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.title("Landscape After 90° Rotation")
plt.xlabel("x")
plt.ylabel("y")

plt.grid(True, alpha=0.3)
plt.axis("equal")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# ORIGINAL → 90° ROTATED POINTS
# ============================================================

plt.figure(figsize=(9, 7))

plt.scatter(
    USER_POINTS[:, 0],
    USER_POINTS[:, 1],
    s=100,
    label="Original"
)

plt.scatter(
    rotated_points[:, 0],
    rotated_points[:, 1],
    s=100,
    label="After 90° Rotation"
)

# Draw transformation arrows
for original, rotated in zip(
    USER_POINTS,
    rotated_points
):

    plt.annotate(
        "",
        xy=rotated,
        xytext=original,
        arrowprops=dict(
            arrowstyle="->",
            linewidth=1.5
        )
    )

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.xlabel("x-coordinate")
plt.ylabel("y-coordinate")

plt.title(
    "Coordinate Transformation Under Orthogonal Rotation"
)

plt.legend()
plt.grid(True, alpha=0.3)
plt.axis("equal")

plt.show()
```

```python
# ============================================================
# SUCCESSIVE ROTATION VISUALIZATION
# ============================================================

points_0 = USER_POINTS
points_90 = rotate_points(USER_POINTS, Q_90)
points_180 = rotate_points(USER_POINTS, Q_180)

plt.figure(figsize=(9, 7))

plt.scatter(
    points_0[:, 0],
    points_0[:, 1],
    s=100,
    label="0°"
)

plt.scatter(
    points_90[:, 0],
    points_90[:, 1],
    s=100,
    label="90°"
)

plt.scatter(
    points_180[:, 0],
    points_180[:, 1],
    s=100,
    label="180°"
)

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.xlabel("x")
plt.ylabel("y")

plt.title("0° → 90° → 180° Screen Rotation")

plt.legend()
plt.grid(True, alpha=0.3)
plt.axis("equal")

plt.show()
```

```python
# ============================================================
# BENCHMARK TEST DATASET
# ============================================================

benchmark_points = np.array([
    [1, 0],
    [0, 1],
    [2, 3],
    [-4, 2],
    [5, -1]
], dtype=float)

# Expected results for 90° rotation:
# (x, y) → (-y, x)

expected_90 = np.array([
    [0, 1],
    [-1, 0],
    [-3, 2],
    [-2, -4],
    [1, 5]
], dtype=float)

benchmark_result = rotate_points(
    benchmark_points,
    Q_90
)

print("=" * 70)
print("BENCHMARK TEST")
print("=" * 70)

all_tests_passed = True

for i, (actual, expected) in enumerate(
    zip(benchmark_result, expected_90),
    start=1
):

    passed = np.allclose(actual, expected)

    if not passed:
        all_tests_passed = False

    print(
        f"Test {i}: "
        f"Input = {benchmark_points[i-1]} | "
        f"Output = {actual} | "
        f"Expected = {expected} | "
        f"PASS = {passed}"
    )

print("\nAll benchmark tests passed:", all_tests_passed)
```

```python
# ============================================================
# FINAL PROJECT VERIFICATION
# ============================================================

orthogonality_test = np.allclose(
    Q_90.T @ Q_90,
    np.eye(2)
)

determinant_test = np.isclose(
    abs(np.linalg.det(Q_90)),
    1
)

distance_test = np.allclose(
    np.linalg.norm(USER_POINTS, axis=1),
    np.linalg.norm(points_90, axis=1)
)

quadratic_test = np.allclose(
    original_q,
    rotated_q
)

double_rotation_test = np.allclose(
    Q_90 @ Q_90,
    Q_180
)

print("=" * 70)
print("FINAL PROJECT VERIFICATION")
print("=" * 70)

print(f"1. QᵀQ = I                  : {'PASS' if orthogonality_test else 'FAIL'}")
print(f"2. |det(Q)| = 1             : {'PASS' if determinant_test else 'FAIL'}")
print(f"3. Distances preserved       : {'PASS' if distance_test else 'FAIL'}")
print(f"4. Quadratic form preserved  : {'PASS' if quadratic_test else 'FAIL'}")
print(f"5. Q(90°)² = Q(180°)        : {'PASS' if double_rotation_test else 'FAIL'}")
print(f"6. Benchmark tests           : {'PASS' if all_tests_passed else 'FAIL'}")

all_verified = all([
    orthogonality_test,
    determinant_test,
    distance_test,
    quadratic_test,
    double_rotation_test,
    all_tests_passed
])

print("\n" + "=" * 70)

if all_verified:
    print("🎉 ALL MATHEMATICAL AND COMPUTATIONAL TESTS PASSED!")
else:
    print("Some tests failed. Check the configuration.")

print("=" * 70)
```
